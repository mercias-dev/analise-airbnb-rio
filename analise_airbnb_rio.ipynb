```python
import pandas as pd

# Usando um conjunto de dados do Rio de Janeiro hospedado no GitHub (sem bloqueios HTTP)
url = "https://gist.githubusercontent.com/eladio123/3760bd95ba4dc48f7925a34d716e41fe/raw/listings.csv"

# Carregando a tabela
df = pd.read_csv(url)

print("Dados carregados com sucesso!")
```

```python
df.head()
```

```python
# Ver o número de linhas e colunas
print(f"O conjunto de dados tem {df.shape[0]} linhas e {df.shape[1]} colunas.\n")

# Ver informações gerais sobre as colunas
df.info()
```

```python
# 1. Média e Mediana de preço
preco_medio = df['price'].mean()
preco_mediano = df['price'].median()

print(f"Preço médio por noite: R$ {preco_medio:.2f}")
print(f"Preço mediano por noite: R$ {preco_mediano:.2f}\n")

# 2. Top 10 Bairros com mais anúncios no Airbnb
print("--- TOP 10 BAIRROS COM MAIS ANÚNCIOS ---")
print(df['neighbourhood'].value_counts().head(10))

# 3. Tipos de acomodação mais comuns
print("\n--- TIPOS DE ACOMODAÇÃO ---")
print(df['room_type'].value_counts())
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Definir o tamanho da imagem do gráfico
plt.figure(figsize=(10, 5))

# Criar o gráfico de barras dos 10 bairros com mais anúncios
top_bairros = df['neighbourhood'].value_counts().head(10)
sns.barplot(x=top_bairros.values, y=top_bairros.index, palette='Blues_r')

# Adicionar títulos e rótulos organizados
plt.title('Top 10 Bairros com Mais Anúncios no Airbnb (Rio de Janeiro)', fontsize=14)
plt.xlabel('Quantidade de Anúncios', fontsize=12)
plt.ylabel('Bairro', fontsize=12)

# Exibir o gráfico
plt.show()
```
